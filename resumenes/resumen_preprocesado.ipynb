## 2. Distribución de la variable objetivo

**Observaciones sobre el desbalanceo:**

> Vemos que hay un claro desbalanceo de clases. La clase "Matriculado" está claramente poco representada en el dataset respecto a "Graduado" que agrupa casi el 50% de los datos

**Estrategia adoptada:**

> Lo trataremos directamente en la implementación de los modelos usando class_weigth = "balanced". Como exploración adicional haremos algunas comparaciones usando SMOTE para generar datos artificales.

---

## 3. Outliers — edad al matricularse

**Análisis de la distribución:**

> La variable que más nos preocupa en cuanto a outliers es la edad al matricularse ya que presenta una clara asimetría que alcanza valores muy elevados prácticamente incoherentes. Que una persona tenga 70 años al matricularse no parece muy realista...

**Decisión sobre los outliers:**

> No vamos a eliminar los outliers de momento y analizaremos que impacto tienen sobre nuestro modelo. Veremos si su eliminación mejores la calidad del modelo usado lo eliminaremos posteriormente

---

## 4. Corrección de asimetría — Yeo-Johnson

**Variables tratadas y resultado:**

> Aplicando la transformación sobre todas las variables continuas del dataset vemos que destacan 3 en específico por tener una asimetría con valor absoluto mayor que 1. Además, algunas de las variables admiten valor 0 y Box Cox no funciona bien con eso. Por ello usaremos la transformaión de Yeo-Johnson.

**Limitaciones observadas:**

> Vemos que en edad al matricularse la corrección genera visualmente una distribución transformada irregular. Eso ha ocurrido por los outliers extremos que tiene provocando que nuestra transformación no arregle el problema y aparezcan dos poblaciones claramente distintas que ninguna transformación consige solucionar. Por tanto DECIDIMOS ELIMINAR LOS OUTLIERS de la variable edad al matricularse. Aunque lo habíamos dejado para el futuro, hemos encontrado ya fallos desde el preprocesado por lo que decidimos eliminarlas directamente desde el principio.

---

## 5. Estandarización — StandardScaler

**Efecto del escalado:**

> Vemos que antes del escalado, las variables presentan escalas muy distintas. Eso haría que algoritmos basados en distancias funcionen mal. Tras aplicar StandartScaler vemos que todas las variables numéricas quedan centradas en media prácticamente 0 y desviación típica también. Vemos que los outliers de edad al matricularse siguen siendo visibles y refuerza la decisión de eliminarlos
---

Trás eliminar los outliers de la variable edad_al_matricularse vemos que ya no necesita una corrección de asimetría por lo que modificamos el código original de preprocesing para adaptarlo a la decisiones tomadas.
