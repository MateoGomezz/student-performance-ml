## 2. Decisiones de Diseño — Clasificación

### 2.1 Regresión Logística

<div style="border: 2px solid #2e86c1; border-radius: 8px; padding: 20px 24px; margin: 28px 0; background-color: #eaf4fb; color: #000000;">
<h3 style="color: #1a5276; margin-top: 0; padding-bottom: 8px; border-bottom: 1px solid #aed6f1; margin-bottom: 14px;">Decisiones de Diseño — Regresión Logística</h3>

Con 98 features el modelo tiene mucho riesgo de sobreajuste. Para ello, vamos a usar regularización para penalizar los coeficientes grandes y reducir la complejidad del modelo. 
Nuestro target tiene tres clases así que usamos la versión multinomial de la regresión logística. La clase predicha es la de mayor probabilidad.
El hiperparámetro clave es c = inversa de la fuerza de regularización y probaremos dos tipos de regularización: Ridge Y Lasso. Para el desbalanceo de clases usamos class_weight = "balanced" que hace que el modelo penalice más los errores sobre la clase minoritaria en el dataset (es decir sobre "Matriculado")
-Para encontrar la regularización que mejor funciona y el hiperparámetro óptimo, haremos cross-validation con k=5 folds sobre el conjunto X_train obtenido en la parte 2 del proyecto. Probaremos 6 valores de c (usando escala logarítmica para cubrir buenos valores) y 2 tipos de regularización. Por tanto, tendremos 12 combinaciones de hiperparámetros por cada fold y por tanto 60 entrenamientos en nuestra fase de búsqueda.
Por cada fold, tendremos el error cometido por cada combinación (c,función_regularizadora) y GridSearchCV coge automáticamente la combinación (c,función_regularizadora)  con menor error medio  (mayor macro F1-score medio) considerando todos los folds.
Finalmente, entrenamos ese modelo sobre todo el conjunto de train (no solo sobre ⅘) y evaluamos una sola vez sobre test. Terminamos generando algunas métricas finales como la matriz de confusión para ver qué tal se ha comportado nuestro modelo.

### 2.2 KNN

<div style="border: 2px solid #2e86c1; border-radius: 8px; padding: 20px 24px; margin: 28px 0; background-color: #eaf4fb; color: #000000;">
<h3 style="color: #1a5276; margin-top: 0; padding-bottom: 8px; border-bottom: 1px solid #aed6f1; margin-bottom: 14px;">Decisiones de Diseño — KNN</h3>

KNN necesita features con misma escala debido a que usa distancias. Esto ya lo tenemos resuelto en nuestro preprocesado con StandardScaler. 
Otra cosa importante es la maldición de la dimensionalidad, al tener 98 features la distancia entre puntos empieza a ser muy similar y por tanto el concepto de vecino "cercano" empieza a perder significado. Esto es una limitación del modelo que es susceptible de provocar peor rendimiento luego cuando estudiemos los resultados en el notebook. 
Por último, otro problema es el  desbalance de clases: KNN es especialmente sensible a esto ya que el voto por mayoría va a favorecer automáticamente a las clases grandes. Para ello, usaremos weights="distance" que pondera el voto de cada vecino inversamente a su distancia y así da algo más de influencia a los vecinos cercanos reduciendo así un poco el efecto del desbalance.
El programa por tanto va a cargar los datos igual que en el modelo anterior y va a ejecutar GridSearchCV con k=5 folds. Los valores de K barridos van a ser [1,3,5,7,11,15,21,31] y los de p van a ser [1,2] obteniendo 16 combinaciones * 5 folds = 80 entrenamientos. Volveremos a usar Macro F1-score como métrica de selección por la mismas razones explicadas en el notebook de regresión logística. Finalmente usaremos el mejor (K,p) obtenido para entrenar sobre todo el train y evaluaremos sobre nuestro conjunto de test para obtener distintas métricas que presentamos y analizamos en el notebook de resumen_knn.ipynb.

### 2.3 Random Forest

<div style="border: 2px solid #2e86c1; border-radius: 8px; padding: 20px 24px; margin: 28px 0; background-color: #eaf4fb; color: #000000;">
<h3 style="color: #1a5276; margin-top: 0; padding-bottom: 8px; border-bottom: 1px solid #aed6f1; margin-bottom: 14px;">Decisiones de Diseño — Random Forest</h3>

Decisiones de diseño:
Los hiperparámetros clave son tres:
n_estimators controlar cuántos árboles vamos a construir: más árboles siempre reduce varianza pero a partir de un punto tiene demasiado coste computacional para no tanta mejora. 
max_features es el número de features candidatas en cada split que sirve de mecanismo de decorrelación y se suele usar sqrt(N_features) = sqrt(98) ≈  10
max_depth controla la profundidad máxima de cada árbol, hay que encontrar equilibrio entre muy sobreajustados o no
Finalmente, para el desbalance de clases usaremos class_weight="balanced" que en Random Forest actúa igual que regresión logística ponderando los errores en cada clase inversamente a su frecuencia durante el ajuste en cada árbol.
El flujo será muy parecido al de modelos anteriores, haremos un GridSearchCV con k= 5 folds barriendo los tres hiperparámetros: n_estimators entre {100,300}, max_features entre {"sqrt",20} y max_depth entre {None,20}, obteniendo 8 combinaciones x 5 folds = 40 entrenamientos. Limitamos las rejillas ya que el tiempo de entrenamiento se hace demasiado largo y hace que la ejecución completa del proyecto sea demasiado larga.
La métrica de selección es Macro F1 por las mismas razones que en los anteriores modelos.

## 4. Decisiones de Diseño — No Supervisado

<div style="border: 2px solid #2e86c1; border-radius: 8px; padding: 20px 24px; margin: 28px 0; background-color: #eaf4fb; color: #000000;">
<h3 style="color: #1a5276; margin-top: 0; padding-bottom: 8px; border-bottom: 1px solid #aed6f1; margin-bottom: 14px;">Decisiones de Diseño — No Supervisado</h3>

Enfoque
Vamos a combinar dos técnicas complementarias. Primero, usaremos PCA (Análisis de Componentes Principales) que nos va a permitir proyectar el espacio original que está en altísima dimensión sobre un subespacio de mucha menor dimensión. Cada columna del nuevo dataset obtenido pasando los datos por esta transformación será la proyección de todos los puntos sobre las direcciones de mayor varianza. Con esto, podremos reducir la dimensionalidad eliminando ruido y colinealidad y sobre todo obtener una representación 2D que nos permita visualizar los datos antes de clusterizar. 

Luego, usaremos K-Means para asignar cada punto al centroide más cercano e iteraremos hasta poder que converga. El punto fuerte de este algoritmo es su simplicidad e interpretabilidad pero tenemos que darle el número de clusters K de antemano. Más adelante explicaremos cómo hemos gestionado este problema. 
Selección de variables
Partimos de X_train.csv, tenemos las 98 columnas del preprocesado. Sin embargo, de las 98 columnas tenemos unas 70 que vienen del One-Hot-Encoding por lo tanto muchas valen 0 para muchísimos estudiantes y 1 solo para algunos que pertenecen a esa categoría en concreto. Por tanto, cuando PCA busca las direcciones de mayor varianza en el espacio conjunto, estas 70 columnas dominan por su volumen y entonces el primer componente principal seguramente acabaría describiendo diferencia en la carrera cursada o en la vía de acceso y no lo que realmente nos interesa: los perfiles académicos y socioeconómicos. En K-Means se extendería el problema: dos estudiantes con rendimientos académicos distintos podrían verse cerca por coincidir en muchas columnas binarias.

La solución es entonces seleccionar variables numéricas continuas con significado real y ya estandarizadas en el preprocesado. El subconjunto de variables seleccionadas cubre tres ejes que de verdad nos permiten diferenciar perfiles de estudiantes:
Rendimiento académico: son las variables con mayor poder discriminativo del dataset y lo vimos con el feature importance en Random Forest y al analizar coeficientes de regresión logística. Excluimos asignaturas_sin_evaluacion y asignaturas_convalidadas ya que tienen mucho 0 y pueden aportar ruido.
Perfil de acceso: guardamos nota_admision, nota_cualifiacion_previa, edad_al_matricularse que son variables que capturan el punto de partida del estudiante antes de entrar en la universidad además de ser numéricas continuas con variabilidad real.
Situación económica: guardamos becado, matricula_al_dia y deudor. Aunque son binarias tienen una semántica propia y no son el resultado de partir una categórica en muchísimos trozos. Permiten describir la situación económica del estudiante. De hecho, matricula_al_dia fue uno de los coeficientes que mantuvo Lasso en regresión por lo que ya veníamos intuyendo que tiene efecto real sobre los resultados.
Excluimos todas las columnas de One Hot Encoding porque no describen al estudiante individualmente o producen un ruido agregado problemático. El resultado es un subconjunto de 14 variables que cubre los ejes fundamentales de diferenciación entre los estudiantes.



Flujo
Ahora sí, ajustamos PCA sobre ese subconjunto de 14 variables. Por otro lado usamos el scree plot que nos permite ver cuántos componentes acumulan la mayor parte de la varianza explicada. Nos quedaremos con el número de componentes adecuado y usaremos PC1 y PC2 para una visualización sencilla en 2D. Analizaremos los loadings de cada componente para ver qué combinación lineal de las variables originales forma cada dirección de máxima varianza.

Luego encontraremos el valor óptimo de K. Para ellos usaremos dos criterios estudiados durante el curso: el método del codo (se centra en cumplir la compactitud dentro de los clusters) y Silhouette (que también mira la separación entre clusters). Probaremos K entre 2 y 8, obtendremos las gráficas de ambos métodos y nos quedaremos con el valor de K que maximiza el Silhouette score debido a que se considera una métrica más completa para hallar un K óptimo. Además, usamos init='k-means++' que permite una inicialización inteligente de los centroides.
